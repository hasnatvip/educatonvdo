# EducationVdo - Lightning AI Studio Launcher
This notebook is specifically configured for Lightning.ai studios. It avoids Colab-specific obfuscations and leverages the clean workspace provided by Lightning.

**Note:** You can access the UI through Lightning Studio`s built-in "Web App" port forwarding on port `7860`, or you can toggle `enable_gradio_share = True` below to generate a public grad.io link.

```python
import os
import subprocess
import torch

# ---------------- CONFIGURATION ----------------
enable_gradio_share = True
# -----------------------------------------------

print("Detecting device...")
if torch.cuda.is_available():
    device = "cuda"
    print("Using GPU (CUDA) - Optimal for Lightning AI")
else:
    device = "cpu"
    print("Using CPU")

# Pre-patch the default layout file to enable Gradio Share if requested
layout_file = "educationvdo/uis/layouts/default.py"
if os.path.exists(layout_file):
    with open(layout_file, "r") as f:
        content = f.read()
    
    if enable_gradio_share:
        if "launch(share=True" not in content:
             content = content.replace("launch(f", "launch(share=True, f")
             with open(layout_file, "w") as f:
                 f.write(content)
             print("Enabled Gradio public share link.")
    else:
        if "launch(share=True" in content:
             content = content.replace("launch(share=True, f", "launch(f")
             with open(layout_file, "w") as f:
                 f.write(content)

cmd = f"python educationvdo.py run --execution-providers {device}"

print(f"\nStarting EducationVdo Command:\n{cmd}\n")
print("If running in Lightning Studio, you can click on the Web App tab to expose port 7860!")
subprocess.run(cmd, shell=True)
```
